```python
from investiny import historical_data, search_assets
import pandas as pd
from talib import abstract
import plotly.graph_objects as go
import plotly.express as px
```

# DATA EXPLORATION

```python
invest_id = search_assets(query="DJI", limit=1, type="Index", exchange="NYSE")[0][
    "ticker"
]
```

## DJI from 1988

```python
checkpoints = [
    ("01/01/1988", "12/31/1998"),
    ("01/01/1999", "12/31/2009"),
    ("01/01/2010", "12/31/2020"),
    ("01/01/2021", None),
]
```

```python
data = pd.DataFrame(
    historical_data(investing_id=invest_id, interval="D", from_date="01/01/2008")
)
```

```python
data
```

```python
fig1 = go.Figure()
trace1 = go.Candlestick(
    x=data["date"], close=data.close, open=data.open, high=data.open, low=data.low
)
fig1.add_traces([trace1])
fig1.show()
```
